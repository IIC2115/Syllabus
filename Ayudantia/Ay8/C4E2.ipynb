# E2

Consultas SQL sobre la base creada en `E1.ipynb`.

```python
import sqlite3

conn = sqlite3.connect("vehicles_e1.db")
cur = conn.cursor()
```

1. Tipo de carrocería y cantidad de vehículos, de mayor a menor.

```python
consulta_1 = """SELECT C.tipo, COUNT(V.veid) AS cantidad
FROM Carrocerias C
LEFT JOIN Vehiculos V ON C.cid = V.carroceria_id
GROUP BY C.cid, C.tipo
ORDER BY cantidad DESC"""

cur.execute(consulta_1)
for fila in cur.fetchall():
    print(fila)
```

2. Cinco vendedores con más ventas.

```python
consulta_2 = """SELECT Ve.nombre, COUNT(V.sid) AS cantidad_ventas
FROM Vendedores Ve
JOIN Ventas V ON Ve.vid = V.vendedor_id
GROUP BY Ve.vid, Ve.nombre
ORDER BY cantidad_ventas DESC
LIMIT 5"""

cur.execute(consulta_2)
for fila in cur.fetchall():
    print(fila)
```

3. Precio promedio por fabricante, considerando los que vendieron al menos 10 vehículos.

```python
consulta_3 = """SELECT F.nombre, AVG(V.sellingprice) AS precio_promedio
FROM Ventas V
JOIN Vehiculos VH ON V.vehiculo_id = VH.veid
JOIN Modelos M ON VH.modelo_id = M.mid
JOIN Fabricantes F ON M.fabricante_id = F.fid
GROUP BY F.fid, F.nombre
HAVING COUNT(DISTINCT VH.veid) >= 10
ORDER BY precio_promedio DESC"""

cur.execute(consulta_3)
for fila in cur.fetchall():
    print(fila)
```

4. Vehículos vendidos por al menos 20% menos que su MMR.

```python
consulta_4 = """SELECT VH.vin, F.nombre, M.nombre, V.sellingprice, V.mmr
FROM Ventas V
JOIN Vehiculos VH ON V.vehiculo_id = VH.veid
JOIN Modelos M ON VH.modelo_id = M.mid
JOIN Fabricantes F ON M.fabricante_id = F.fid
WHERE V.sellingprice <= V.mmr * 0.8
ORDER BY V.sellingprice"""

cur.execute(consulta_4)
resultado_4 = cur.fetchall()
print("Total de filas:", len(resultado_4))
print("Primeras 10 filas:")
for fila in resultado_4[:10]:
    print(fila)
```

5. Cantidad de ventas y odómetro promedio por estado, de mayor a menor cantidad.

```python
consulta_5 = """SELECT state, COUNT(*) AS cantidad_ventas, AVG(odometer) AS odometro_promedio
FROM Ventas
GROUP BY state
ORDER BY cantidad_ventas DESC"""
cur.execute(consulta_5)
for fila in cur.fetchall():
    print(fila)
```

6. Fabricante con mayor variedad de modelos vendidos.

```python
consulta_6 = """SELECT F.nombre, COUNT(DISTINCT M.mid) AS modelos_unicos
FROM Ventas V
JOIN Vehiculos VH ON V.vehiculo_id = VH.veid
JOIN Modelos M ON VH.modelo_id = M.mid
JOIN Fabricantes F ON M.fabricante_id = F.fid
GROUP BY F.fid, F.nombre
ORDER BY modelos_unicos DESC
LIMIT 1"""
cur.execute(consulta_6)
print(cur.fetchone())
conn.close()
```
