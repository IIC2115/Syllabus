# Modelos predictivos basados en aprendizaje de máquina

Uno de los objetivos principales del análisis de datos es la construcción de modelos predictivos basados en datos, que permitan inferir el comportamiento de un fenómeno en escenarios nuevos o en el futuro. En este capítulo, nos centraremos en técnicas de aprendizaje de máquina para construir estos modelos predictivos. Para esto, utilizaremos la librería `scikit-learn`, que es una de las más utilizadas para este tipo de análisis.

Antes de describir los modelos y el uso de la librería, presentaremos una breve introducción a los conceptos teóricos esenciales del aprendizaje de máquina. En particular, describiremos los distintos tipos de aprendizaje, el mecanismo de entrenamiento de un modelo y la forma de elegir y evaluar modelos mediante conjuntos de entrenamiento, validación y test, además de métricas de rendimiento.


## ¿Qué es el aprendizaje de máquina?

El aprendizaje de máquina, o _Machine Learning_ (ML) en inglés, es una disciplina que combina conocimientos y técnicas de ciencia de la computación, cálculo y estadística, donde se busca construir modelos de análisis y procesamiento de datos que permitan resolver problemas específicos. Estos modelos cumplen en general las siguientes características:
* Modelos (principalmente) predictivos basados en datos, enfocados en problemas específicos.
* Mejoran con la experiencia/datos (generalmente, mientras más, mejor).
* Buscan aprendizaje más que modelamiento de datos.

Estas características, unidas a su buen rendimiento en problemas reales y a la gran cantidad de librerías relacionadas disponibles, hacen que las técnicas de _machine learning_ sean ampliamente usadas para analizar datos. Algunos ejemplos de uso del aprendizaje de máquina son los detectores automáticos de fraude bancario, los filtros anti-spam, los sistemas recomendadores y los sistemas de traducción entre idiomas.

## Tipos de aprendizaje de máquina

Si bien es posible definir más tipos, en general las técnicas y modelos de _machine learning_ pueden corresponder a tres tipos distintos de aprendizaje: supervisado, no supervisado y reforzado:

<img src="figs/ml_categories.png" width="400px" align="center"/>

El aprendizaje supervisado es el más común e incluye problemas como regresión y clasificación. En este esquema, el objetivo es predecir para cada dato ya sea un valor numérico (regresión) o una etiqueta (clasificación). Para lograr esto, se cuenta con un conjunto de datos donde estos valores son conocidos, a partir de los cuales estima un modelo predictivo que idealmente sea capaz de generalizar a datos no vistos anteriormente. Por ejemplo, si los datos consisten en información de las condiciones meteorológicas, incluida la temperatura, para todos los días del año 2020, podríamos construir un modelo de aprendizaje supervisado que, dadas las condiciones meteorológicas de cualquier día del año 2021, prediga la temperatura para el día siguiente.

El aprendizaje no supervisado se centra en el descubrimiento de estructuras o relaciones en los datos, sin contar con etiquetas o valores a predecir, como en el aprendizaje supervisado. Ejemplos de tareas de aprendizaje no supervisado son el _clustering_ y la reducción de dimensionalidad.

Finalmente, el aprendizaje reforzado considera la existencia de un *agente* que debe interactuar con un *entorno*, con el fin de cumplir una tarea. A diferencia del aprendizaje supervisado, no existen etiquetas que indiquen lo que se debe hacer, sino que el agente recibe una recompensa (relacionada con la tarea) al interactuar con el ambiente, y así este buscará maximizar esta recompensa en el tiempo. Ejemplos de dominios donde se aplica el aprendizaje reforzado son el control de personajes en juegos o los procesos de decisión con incerteza.

En este curso nos centraremos en los modelos y técnicas de aprendizaje supervisado y no supervisado. ## Entrenamiento de modelos

Como norma general, los modelos de _machine learning_ trabajan sobre datos multidimensionales, es decir, cada dato es considerado como un vector donde cada dimensión corresponde a un atributo o variable. Por ejemplo, si tenemos un `DataFrame`, podemos considerar que cada fila es un vector y que cada columna representa una de sus dimensiones. Los modelos también pueden trabajar con otros tipos de datos multidimensionales, como imágenes o grafos. Para referirse al espacio vectorial donde viven los datos, generalmente se utiliza el término **espacio de características**, o _feature space_.

La estimación de un modelo se realiza mediante un proceso llamado **entrenamiento**. En él, el modelo aprende a realizar predicciones a partir de ejemplos donde conocemos la respuesta correcta. Para muchos modelos, este aprendizaje se plantea como un problema de optimización: se buscan valores que minimicen una función de pérdida o costo, que mide la discrepancia entre las predicciones y los valores observados.

### Parámetros e hiperparámetros

Entrenar un modelo no solo consiste en aprender valores a partir de los datos. En la práctica también debemos decidir **qué modelo utilizar y cómo configurarlo**. Conviene distinguir dos conceptos:

* Los **parámetros** se aprenden durante el entrenamiento. Por ejemplo, una regresión aprende sus coeficientes y un árbol aprende qué variables y umbrales utilizar en sus divisiones.
* Los **hiperparámetros** son decisiones que tomamos antes de entrenar. Por ejemplo, en KNN debemos elegir el número de vecinos $K$, mientras que en un árbol podemos decidir su profundidad máxima.

Como no conocemos de antemano qué alternativa funcionará mejor, necesitamos datos que permitan comparar configuraciones sin comprometer la evaluación final.

### Entrenamiento, validación y test

Para separar el aprendizaje, la toma de decisiones y la evaluación final, dividimos los datos en tres conjuntos disjuntos:

| Conjunto | Rol |
|---|---|
| **Entrenamiento** | Aprender los parámetros del modelo. |
| **Validación** | Comparar modelos e hiperparámetros y elegir una solución. |
| **Test** | Estimar, una sola vez y al final, el rendimiento de la solución elegida sobre datos no vistos. |

El conjunto de test no debe utilizarse para elegir el modelo. Si probamos varias configuraciones y escogemos la que obtiene mejor rendimiento en test, entonces estamos usando indirectamente ese conjunto para ajustar nuestra solución. El resultado final dejaría de ser una estimación verdaderamente independiente.

Una división frecuente es destinar aproximadamente 60% de los datos a entrenamiento, 20% a validación y 20% a test. Estas proporciones no son reglas universales: dependen de la cantidad de datos y del problema, pero hacen visible el rol distinto de cada conjunto.

### Generalización, _underfitting_ y _overfitting_

Buscamos que el modelo **generalice**, es decir, que funcione bien en casos que no participen del entrenamiento. Un modelo puede sufrir **sobreajuste** u _overfitting_ cuando se adapta demasiado a los ejemplos de entrenamiento: obtiene allí un rendimiento muy alto, pero empeora al enfrentar datos nuevos. Esto puede ocurrir, por ejemplo, si el modelo es más complejo de lo necesario. En el extremo opuesto, existe **subajuste** o _underfitting_, donde la capacidad del modelo no es suficiente para capturar el fenómeno a predecir.

También podemos sobreajustar nuestras **decisiones** al conjunto de validación si probamos demasiadas alternativas y elegimos repetidamente según sus resultados. Por esta razón, aunque validación se utilice varias veces durante la selección, test se mantiene intacto hasta el final.

## Técnicas de aprendizaje supervisado

Como se mencionó anteriormente, existen múltiples modelos de aprendizaje supervisado, cada uno con sus características propias, ventajas y desventajas que los hacen más o menos adecuados a los disintos tipos de problemas de datos que encontremos. A continuación se describirán las características de algunos de ellos.

### K-vecinos cercanos (KNN)
Este algoritmo consiste en realizar predicciones basadas unicamente en los K elementos más cercanos al dato analizado, en el espacio de características. Si la tarea es una clasificación, el resultado se entrega a través de una votación de mayoría (las clase más repetida entre los K vecinos gana) o un promedio si la tarea es una regresion (promedio de los K vecinos). Al solo basarse en el cálculo de distancias, este modelo no necesita entrenamiento, lo que lo hace muy simple y rápido de utilizar. Sin embargo, su rendimiento en cuanto a tiempo de ejecución sufre mucho cuando la cantidad de ejemplos en el set de entrenamiento es alta (piense en la complejidad computación de calcular la distancia euclidiana entre el punto analizado y todos los ejemplos de entrenamiento y luego ordenarlas). El valor de $K$ es un hiperparámetro: no se aprende automáticamente a partir de los datos, sino que debemos elegirlo. Valores muy pequeños pueden producir modelos demasiado sensibles a ejemplos particulares, mientras que valores muy grandes pueden suavizar en exceso las diferencias entre clases.

### Regresión lineal, logística y polinomial
Estas técnicas de aprendizaje estadístico permiten estimar funciones (reg. lineal) o clasificar (reg. logística) en base a una combinación lineal de las características de los datos. Además, si se expanden las características utilizando una transformación polinomial de estas, es posible hacer estimaciones de funciones y clasificación no lineal. Estas técnicas son ampliamente utilizadas debido a su rapidez, sencillez e interpretabilidad, pero tienen problemas para obtener buenos rendimientos en problemas complejos.

### Support Vector Machine
Un Support Vector Machines (SVM) permite hacer clasificación binaria o multiclase, utilizando el concepto de máximo margen. En él, se busca el clasificador lineal (hiperplano) que maximice la distancia entre las clases. Esto permite obtener un poder de generalización más alta que otros métodos como la regresión lineal. Además del rendimiento, los SVM son muy rápidos de entrenar, lo que los hace ser candidatos perfectos cuando se tiene gran volumen de datos.

<img src="figs/svm.png" width="250px" align="center"/>

### Árboles de decisión y regresión
Los árboles de decisión son técnicas simples de clasificación o regresión, que funcionan con cualquier tipo de atributo (numéricos, categóricos, binarios, etc). Su funcionamiento se basa en la construcción  de una estructura de árbol, en base a tests/umbrales calculados para cada característica analizada. Este árbol es construido hasta llegar a los nodos hoja, donde se entrega la predicción o clasificación. La profundidad máxima es un ejemplo de hiperparámetro: limitarla permite controlar la complejidad del árbol. Si bien estos modelos pueden sufrir serios problemas de _overfitting_, su alto nivel de explicabilidad los hace siempre buenos candidatos, sobre todo si se tiene gran cantidad de datos.

<img src="figs/decision_tree.png" width="300px" align="center"/>

### Ensambles de árboles
Lás técnicas de ensamble se basan en la combinación de múltiples árboles para generar una predicción más robusta y menos sesgada. Si bien existen múltiples existen diversas maneras de combinar estas predicciones, las estrategias paralela aleatoria (_Random Forest_) y secuencial (_Gradient Boosting_) son las que mejores resultados obtienen en la práctica. Estas técnicas son en la actualidad de las más competiticas en cuanto a rendimiento al utilizar datos tabulados.

<img src="figs/ensemble.png" width="330px" align="center"/>

### Redes Neuronales
Las redes neuronales son modelos altamente generales y complejos para estimar funciones de cualquier tipo (de hecho, es posible demostrar que pueden aprender cualquier función). Su principal característica es que procesan los datos a través de varias capas, que combinan operaciones lineales y no lineales. Si se tienen muchos datos, estos modelos son por lejos los que mejor funcionan, en particular para datos no estructurados (imágenes, texto, audio, video, etc.)

## Uso de aprendizaje de máquina en Python

Como se mencionó anteriormente, utilizaremos la librería `scikit-learn` para aplicar técnicas de ML al procesamiento de datos. La popularidad de esta librería radica principalmente en el uso de una interfaz limpia, uniforme y simple, que facilita la exploración y permite la integración con otras librerías, como Pandas. En otras palabras, los modelos y técnicas disponibles utilizan una interfaz de ejecución y configuración muy similar. Esto permite obviar parcialmente los detalles teóricos y concentrarse en la exploración y construcción de soluciones.

La librería ya se encuentra instalada en Colab. En otros entornos puede instalarse mediante `pip install scikit-learn`.


### Esquema de representación de datos

El esquema de representación de datos de `scikit-learn` es sumamente directo y sencillo. Salvo algunos casos particulares, todos los modelos reciben como entrada una matriz de _features_ y un vector objetivo. La matriz de _features_, o **X**, tiene dimensiones `n_ejemplos × n_features` y puede representarse mediante un `DataFrame` o un `array` de NumPy. El vector objetivo, o **y**, contiene los valores a predecir para cada ejemplo, tiene `n_ejemplos` elementos y puede representarse mediante una `Series` o un `array` de NumPy.

La matriz **X** puede corresponder al conjunto de entrenamiento, validación o test, pero mantiene siempre el mismo formato y las mismas columnas.

<img src="figs/sklearn_data_model.png" width="400px" align="center"/>

El siguiente ejemplo muestra cómo cargar un conjunto de datos y crear conjuntos independientes para entrenamiento, validación y test. Utilizaremos el conjunto de datos [`Iris`](https://archive.ics.uci.edu/ml/datasets/iris). Este conjunto es ideal para practicar conceptos de ML debido a su tamaño y simplicidad: contiene 150 ejemplos pertenecientes a 3 especies de la familia Iris, descritos mediante 4 características. El archivo `iris.csv` debe estar ubicado en la misma carpeta que el notebook.


```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
%matplotlib inline

iris_dataset = pd.read_csv('iris.csv')
iris_dataset.head()
```

Antes de construir los modelos predictivos, es necesario identificar la matriz **X** y el vector **y**. Además, `scikit-learn` requiere que las variables entregadas al modelo sean numéricas. En este ejemplo, `species` es la variable objetivo y sus categorías se codificarán como números naturales.

Esta transformación solo asigna un código a cada etiqueta de la variable objetivo. Las transformaciones que aprendan información desde las _features_, como una estandarización, se ajustarán más adelante **solo con el conjunto de entrenamiento**.


```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
iris_dataset['species'] = label_encoder.fit_transform(iris_dataset['species'])

iris_dataset.dtypes
```

A continuación, se utiliza `train_test_split` dos veces. Primero se reserva 40% de los datos en un conjunto temporal; luego ese conjunto se divide en partes iguales para validación y test. El resultado es una división aproximada 60/20/20.

Usaremos `stratify` para conservar aproximadamente la proporción de las tres especies en cada conjunto y `random_state` para que la división sea reproducible. A partir de este punto, el conjunto de test permanecerá intacto hasta la evaluación final.


```python
training_set, temporary_set = train_test_split(
    iris_dataset.copy(),
    test_size=0.40,
    random_state=42,
    stratify=iris_dataset['species']
)

validation_set, test_set = train_test_split(
    temporary_set,
    test_size=0.50,
    random_state=42,
    stratify=temporary_set['species']
)

print(f'Tamaño set entrenamiento: {len(training_set)}')
print(f'Tamaño set validación: {len(validation_set)}')
print(f'Tamaño set test: {len(test_set)}')
```

Es posible, y muchas veces recomendable, realizar una etapa de preprocesamiento antes de entregar los datos a los modelos. Por ejemplo, es común estandarizar cada _feature_ para que todas se encuentren en una escala comparable.

Este proceso requiere cuidado para evitar **data leakage**, que consiste en incorporar indirectamente información de validación o test durante el entrenamiento. Por eso, el escalador se ajusta con `fit_transform` **solo sobre entrenamiento**. Luego, sin volver a estimar medias ni desviaciones, se aplica `transform` a validación y test.


```python
from sklearn.preprocessing import StandardScaler

features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
target = 'species'

scaler = StandardScaler()
training_set[features] = scaler.fit_transform(training_set[features])
validation_set[features] = scaler.transform(validation_set[features])
test_set[features] = scaler.transform(test_set[features])

training_set.head()
```

### Selección de hiperparámetros con validación

Con los datos preparados, comenzaremos con KNN porque el efecto de su hiperparámetro $K$ es intuitivo. Probaremos varios valores, entrenaremos cada alternativa con el conjunto de entrenamiento y mediremos su _accuracy_ sobre validación.

La métrica _accuracy_ corresponde a la proporción de predicciones correctas. Cuando las clases están desbalanceadas, puede ser más apropiado utilizar _balanced accuracy_ u otras métricas. Lo importante en esta etapa es que **ninguna alternativa consulte el conjunto de test**.


```python
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier

k_values = [1, 3, 5, 7, 9, 11]
k_results = []

for k in k_values:
    knn_model = KNeighborsClassifier(n_neighbors=k)
    knn_model.fit(training_set[features], training_set[target])

    validation_predictions = knn_model.predict(validation_set[features])
    validation_accuracy = accuracy_score(
        validation_set[target], validation_predictions
    )
    k_results.append({'K': k, 'accuracy_validacion': validation_accuracy})

k_results_df = pd.DataFrame(k_results)
k_results_df
```

Elegiremos el valor de $K$ con mayor _accuracy_ de validación. Si hay un empate, preferiremos el mayor valor de $K$ entre las alternativas empatadas, porque corresponde a una frontera de decisión menos flexible. La regla de desempate queda definida antes de mirar test.


```python
best_validation_accuracy = k_results_df['accuracy_validacion'].max()
best_k = int(
    k_results_df.loc[
        k_results_df['accuracy_validacion'] == best_validation_accuracy,
        'K'
    ].max()
)

print(f'Mejor valor de K según validación: {best_k}')
```

### Comparación de familias de modelos

La validación no solo permite elegir hiperparámetros dentro de una familia. También permite comparar familias de modelos. A continuación compararemos el KNN recién configurado con un árbol de decisión y un SVM. Cada alternativa se entrena con el mismo conjunto de entrenamiento y se evalúa con el mismo conjunto de validación.

En una aplicación real también podríamos validar los hiperparámetros del árbol y del SVM. Aquí utilizaremos configuraciones simples para mantener el foco en el procedimiento: todas las decisiones se toman con validación y test continúa reservado.


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

candidate_models = {
    f'KNN (K={best_k})': KNeighborsClassifier(n_neighbors=best_k),
    'Árbol (profundidad máxima=3)': DecisionTreeClassifier(
        max_depth=3, random_state=42
    ),
    'SVM': SVC()
}

family_results = []

for model_name, model in candidate_models.items():
    model.fit(training_set[features], training_set[target])
    validation_predictions = model.predict(validation_set[features])
    validation_accuracy = accuracy_score(
        validation_set[target], validation_predictions
    )
    family_results.append({
        'modelo': model_name,
        'accuracy_validacion': validation_accuracy
    })

family_results_df = pd.DataFrame(family_results)
family_results_df
```

La solución seleccionada será la que obtenga mayor _accuracy_ de validación. Nuevamente, si existe un empate se conservará la primera alternativa. Con esto terminan las decisiones de modelamiento.


```python
best_model_index = family_results_df['accuracy_validacion'].idxmax()
best_model_name = family_results_df.loc[best_model_index, 'modelo']

print(f'Modelo seleccionado: {best_model_name}')
```

### Evaluación final sobre test

Recién ahora utilizamos el conjunto de test. Entrenamos nuevamente la solución seleccionada con el conjunto de entrenamiento y calculamos su rendimiento final **una sola vez**. Este valor entrega una estimación independiente de cómo se comportará la solución elegida frente a datos nuevos.


```python
from sklearn.base import clone

final_model = clone(candidate_models[best_model_name])
final_model.fit(training_set[features], training_set[target])

test_predictions = final_model.predict(test_set[features])
test_accuracy = accuracy_score(test_set[target], test_predictions)

print(f'Accuracy final en test: {test_accuracy:.2f}')
```
